Las soluciones de mayor puntuación en la competición original combinaron típicamente: (a) ingeniería de variables de **retardo temporal** (*lag features*: qué productos tenía el cliente 1, 2 o N meses atrás, indispensable para identificar altas "nuevas"); y (b) modelos independientes por producto en vez de un único modelo multietiqueta. Ninguna de las dos estrategias es aplicable directamente a nuestro extracto de un único mes, lo cual refuerza la limitación señalada en la conclusión general (pregunta 20) sobre la necesidad de recuperar la dimensión temporal para replicar el objetivo original.
